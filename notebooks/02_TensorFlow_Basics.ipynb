```python
#!pip install tensorflow

import tensorflow as tf

print("TensorFlow Version:", tf.__version__)

a = tf.constant(5)
b = tf.constant(10)

print(a)
print(b) 

print("Addition:", tf.add(a, b))
print("Subtraction:", tf.subtract(a, b))
print("Multiplication:", tf.multiply(a, b))
print("Division:", tf.divide(a, b))
```

```python
x = tf.constant([[1, 2], [3, 4]])
y = tf.constant([[5, 6], [7, 8]])

print("Matrix Addition:", tf.add(x, y))
print("Matrix Multiplication:", tf.matmul(x, y))

print("Shape of x:", x.shape)
print("Data type of x:", x.dtype)
```

```python
t = tf.constant([1, 2, 3, 4])

print("First element of t:", t[0])
print("Slice of t:", t[1:3])

reshaped = tf.reshape(t, (2, 2))
print("Reshaped Tensor:\n", reshaped)

rand = tf.random.uniform((2, 2))
print("Random Tensor:\n", rand)

c = tf.cast(a, tf.float32)
print("Type casted tensor:", c)
```

```python
print("Sum:", tf.reduce_sum(t))
print("Mean:", tf.reduce_mean(t))
print("Max:", tf.reduce_max(t))
print("Min:", tf.reduce_min(t))
```
